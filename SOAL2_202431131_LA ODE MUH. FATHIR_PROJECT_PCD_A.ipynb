## IMPORT LIBRARY

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## MEMBACA GAMBAR


```python
img = cv2.imread('aset21.jpeg')
```

## KONVERSI BGR to RGB


```python
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

## KONVERSI to GRAYSCALE

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

## OPERASI PENCERAHAN

```python
beta = 80
gray_brightened = np.clip(gray.astype(np.int16) + beta, 0, 255).astype(np.uint8)
```

## OPERASI KONTRAS


```python
alpha = 1.5
gray_contrasted = np.clip(alpha * gray.astype(np.float32), 0, 255).astype(np.uint8)
```

## OPERASI GABUNGAN

```python
alpha_gab = 1.3
beta_gab = 60
gray_final = np.clip(alpha_gab * gray.astype(np.float32) + beta_gab, 0, 255).astype(np.uint8)
```

## MENAMPILKAN HASIL

### CITRA ASLI

```python
plt.figure(figsize=(15, 5), dpi=150)
plt.subplot(1, 2, 1)
plt.imshow(rgb)
plt.title("Gambar Asli")
plt.axis('on')

plt.subplot(1, 2, 2)
plt.hist(rgb.ravel(), 256, [0, 256], color='blue')
plt.title("Histogram Gambar Asli")
plt.show()
```

### CITRA GRAY

```python
plt.figure(figsize=(15, 5), dpi=150)
plt.subplot(1, 2, 1)
plt.imshow(gray, cmap='gray')
plt.title("Gambar Gray")
plt.axis('on')

plt.subplot(1, 2, 2)
plt.hist(gray.ravel(), 256, [0, 256], color='gray')
plt.title("Histogram Gambar Gray")
plt.show()
```

## CITRA GRAY YANG DIPERCERAH

```python
plt.figure(figsize=(15, 5), dpi=150)
plt.subplot(1, 2, 1)
plt.imshow(gray_brightened, cmap='gray')
plt.title("Gambar Gray yang Dipercerah")
plt.axis('on')

plt.subplot(1, 2, 2)
plt.hist(gray_brightened.ravel(), 256, [0, 256], color='orange')
plt.title("Histogram Gambar Gray yang Dipercerah")
plt.show()
```

## CITRA GRAY YANG DIPERKONTRAS

```python
plt.figure(figsize=(15, 5), dpi=150)
plt.subplot(1, 2, 1)
plt.imshow(gray_contrasted, cmap='gray')
plt.title("Gambar Gray yang Diperkontras")
plt.axis('on')

plt.subplot(1, 2, 2)
plt.hist(gray_contrasted.ravel(), 256, [0, 256], color='green')
plt.title("Histogram Gambar Gray yang Diperkontras")
plt.show()
```

### CITRA GRAY YANG DIPERCERAH DAN DIPERKONTRAS

```python
plt.figure(figsize=(15, 5), dpi=150)
plt.subplot(1, 2, 1)
plt.imshow(gray_final, cmap='gray')
plt.title("Gambar Gray yang Dipercerah dan Diperkontras")
plt.axis('on')

plt.subplot(1, 2, 2)
plt.hist(gray_final.ravel(), 256, [0, 256], color='purple')
plt.title("Histogram Gambar Gray yang Dipercerah dan Diperkontras")
plt.show()
```
